````python
Here are several ways to write a function to add two numbers in JavaScript:

1. Basic function declaration:
```javascript
function addNumbers(a, b) {
    return a + b;
}
```

2. Arrow function:
```javascript
const addNumbers = (a, b) => {
    return a + b;
};
```

3. Concise arrow function:
```javascript
const addNumbers = (a, b) => a + b;
```

4. Function with type checking:
```javascript
function addNumbers(a, b) {
    // Ensure inputs are numbers
    if (typeof a !== 'number' || typeof b !== 'number') {
        throw new Error('Both inputs must be numbers');
    }
    return a + b;
}
```

5. Function with default parameters:
```javascript
function addNumbers(a = 0, b = 0) {
    return a + b;
}
```

Example usage:
```javascript
console.log(addNumbers(4, 3));  // Output: 7
console.log(addNumbers(10, 5)); // Output: 15
```

Choose the style that best fits your coding preferences and project requirements.
````
